```python
import json
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge

DATA = next((p for p in (Path('data'), Path('../data')) if (p/'real_estate.csv').is_file()), None)
if DATA is None:
    raise FileNotFoundError('Mở notebook mẫu từ thư mục py30 hoặc bai_lam.')
real_estate = pd.read_csv(DATA / 'real_estate.csv')

def _check(name, fn):
    try:
        ok, detail = fn()
    except NotImplementedError:
        ok, detail = False, 'Chưa cài đặt (NotImplementedError)'
    except Exception as error:
        ok, detail = False, f'{type(error).__name__}: {error}'
    return {'name': name, 'ok': bool(ok), 'detail': str(detail)}

def _report(ex_id, *checks):
    print('__PY30__' + json.dumps({'id': ex_id, 'checks': list(checks)}, ensure_ascii=False))
def _partition(fn):
    parts = fn(20,4,3,7)
    return list(map(len,parts)) == [13,4,3] and np.array_equal(np.sort(np.concatenate(parts)), np.arange(20))
def _no_leak(fn):
    X=np.array([[1.,4.],[3.,4.],[101.,4.],[1001.,4.]])
    a,b,c,mu = fn(X,[0,1],[2],[3])
    return np.allclose(mu,[2,4]) and np.allclose(a,[[-1,0],[1,0]]) and np.allclose(b,[[99,0]]) and np.allclose(c,[[999,0]])

def _shuffled(fn):
    a = np.concatenate(fn(40, 8, 8, 1)); b = np.concatenate(fn(40, 8, 8, 2))
    return not np.array_equal(a, np.arange(40)) and not np.array_equal(a, b)
```
